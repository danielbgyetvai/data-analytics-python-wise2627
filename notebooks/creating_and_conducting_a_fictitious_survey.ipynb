Defining the `create_survey()` function

```python
# max_questions is an argument 
# of the function create_survey() 
# here with the default value 10
def create_survey(max_questions = 10):
    # creating an empty list
    questions = []
    # iteratively collect questions using input()
    # i is counter/index of the questions
    for i in range(max_questions):
        # let the user enter the text of the question
        question = input(' Enter the next question: ')
        # let us add the question 
        # to the list of questions
        # if empty string (left blank)
        if question == '':
            # then simply break the loop
            break
        questions.append(question)
    return questions
```

```python
survey = create_survey() 
```

```python
def respond(survey):
    # empty list of answers
    answers = []
    # collecting the answers
    for (i, question) in enumerate(survey):
        answer = input('Question ' + str(i) + ' ' + question + ' ')
        answers.append(answer)
    return answers
```

```python
response = respond(survey)
```

### II. Using Classes

Motivation:

this example is about getting familiar with object oriented programming in Python

to get some understanding of what happens when we write:

`dataframe.columns` (columns attribute of the dataframe object for example) 
or `dataframe.sort_values()` (calling function/method sort_values() with the dataframe object)



```python
# class of objects 
# with some common "behaviour" / types of attributes
# for example in pandas DataFrame is a class
class Survey:
    # functions that objects of the Survey class should have
    # initialiser function/method
    # when an object of the survey class is being created
    # this function is executed
    # opportunity to specify/initialise some attributes
    
    # by convention that the first argument is called self
    # self refers to the object itself

    def __init__(self):
        # let us just have an empty list of questions at the beginning / when the particular survey object is created
        self.questions = []
        # let us just have an empty list of responses at the beginning / when the particular survey object is created
        self.responses = []


    # add_questions() function / method
    # with second argument max_questions (with default value 10)
    def add_questions(self, max_questions=10):        
        # looping - executing the same code multiple times with possibly different values
        # objective of the for loop: set an upper bound for the number questions that can be entered in one run
        for i in range(max_questions):
            # at each iteration
            # let the user enter a new question
            # call the text object entered by the user new_question
            new_question = input()
            # if the user did not enter anything
            if new_question == '':
                # then let the loop end immediately
                break
            # otherwise / if the loop has not ended yet
            # add the new question entered by the user to the list of questions
            self.questions.append(new_question)

        # by default: the add_questions() function will ask the user ten times to enter a new question 
    
    # function called respond()
    # to collect responses
    # response = list of answers from one respondent
    # responses = list of responses
    # responses = list of lists of answers by each respondent 
    # because the function belongs to the class, need to have self/object itself (survey object itself here) 
    # as argument
    def respond(self):
        # empty list to which the answers of a respondent are added
        response = [] 
        # for each question collect an answer 
        # a for loop may be appropriate
        # iterating over elements of a list is possible as well
        # for each question in the list of questions belonging to the survey
        for question in self.questions:
            # collect an answer
            # interactively with input()
            # input(question) makes sure the question is displayed
            answer = input(question)
            # add the answer to the list of answers, called response
            response.append(answer)
        # adding the response to the list of responses belonging to the survey object
        self.responses.append(response)


```

```python
class Survey:
    def __init__(self):
        self.questions = []
        self.responses = []

    def add_questions(self, max_questions=10):        
        for i in range(max_questions):
            new_question = input()
            if new_question == '':
                break
            self.questions.append(new_question)
```
