With the `value_counts()` method, one can compute 

 - frequencies (number of occurrences) 
 - and relative frequencies (proportions)

for distinct values / value-combinations of variables

for a single variable: how often a particular value occurs

for multiple variables: how often a particular combination of values occurs



```python
# frequencies / number observations for each value
# one way
df['year'].value_counts(
    normalize=False
)
# another way
df.value_counts(
    subset=['year'],
    normalize=False
)

```

```python
# at this point the resulting object is a single column 
# (Series in pandas teminology)
type(
    df.value_counts(
        subset=['year'],
        normalize=False
    )   
)

# with the values in the column being the frequencies / number of occurrences
# and the index (on the left) being the distinct values of the variable
```

```python
# to make the index into a column
# reset_index() --> make the index into a column 
df.value_counts(
    subset=['year'],
    normalize=False
).reset_index()
```

### Unit of observation

What does each row describe?
 - Does each row describe a separate / distinct person?
 - Does each row correspond to information on a person in a given year?
 - Does each row correspond to information on a person in a given month?

```python
df
```

Based on visual inspection, it seems to be the case that the unit of observation is `person_id` X `year_month`. According to this hypothesis: each observation would contain information on a given individual in a given month.

How can we test that this happens to be the case?
How can we test that the combination of `person_id` and `year_month` uniquely identifies observations/rows?



```python
# ruling out technical duplicates 
# in order not to have exactly the same column twice
# let us just use drop_duplicates()
# this will make sure that if two rows are fully identical
# then only one is kept
df = df.drop_duplicates()
```

```python
# using value_counts() to test 
# whether a specific combination of variables
# uniquely identifies the observations 
```

```python
# let us first count the number observations
# per combination of person_id and year_month
df.value_counts(
    subset=['person_id', 'year_month']
).reset_index()
# for each pair of person_id X year_month
# count corresponds to the number of observations


```

```python
# let us count how many times 
# a person_id X year_month pair occurs once, twice, three times etc
# simply call a second value_counts() 
# on the frequencies / number of occurrences
(
    df
        .value_counts(
            subset=['person_id', 'year_month']
            )
        .reset_index()
        .value_counts(
            subset=['count']
        )
)
```

```python

# person_id and year do not uniquely identify observations
(
    df
        .value_counts(
            subset=['person_id', 'year']
            )
        .reset_index()
        .value_counts(
            subset=['count']
        )
)
```

### Employment patterns around major events

```python
df2 = df.copy()
```

```python
# between is inclusive: both bounds are included
# adding another condition: & for 'and' and | for 'or'
# this is pandas specific meaning of & and |
# in Python: and, or

# reset_index(drop=True) because only specific/a subset of rows are kept
# this is optional
df2 = df2.loc[df2['year'].between(left=2019,right=2020) &
        df2['sample'].eq('Cross-sectional'),:].reset_index(drop=True)
```
