##### Lists and Tuples
After creating the object
 - you can modify a list
 - you cannot modify a tuple

```python
# creating a list with information on an economist
an_economist = ['Joan Robinson', 1903] 
print(an_economist)
```

```python
# year of death
an_economist.append(1983)
# country of birth
an_economist.append('UK')
# nobel_prize_year
an_economist.append(1975)
```

```python
an_economist
```

Lists can be changed after creation

```python
# indexing in Python starts with 0
# in contrast to R (in R, indexing starts with 1)
an_economist[4] = False
```

```python
an_economist
```

Tuple with information on another economist

```python
another_economist = 'Ragnar Frisch', 1895, 1973, 'Norway', 1969
another_economist
```

```python
# elements of a tuple cannot be changed after creation
another_economist[0] = 'Ragnar Anton Kittil Frisch'
```

```python
# converting 
# a list (here the one containing information on Joan Robinson) 
# to a tuple
an_economist = tuple(an_economist)
```

```python
# checking the type of our object 
type(an_economist)
```

# Dictionaries

- unambiguous assignment of *values* to *keys*
- each *key* occurs only once


```python
# creating a dictionary 
# with information on yet another economist 
# [] lists, () tuples, {} dicts or sets
# empty dictionary
yet_another_economist = {}

# adding items to the dictionary
yet_another_economist['name'] = 'John Maynard Keynes'
yet_another_economist['year_of_birth'] = 1883
yet_another_economist['year_of_death'] = 1946
yet_another_economist['country_of_birth'] = 'UK'
yet_another_economist['nobel_prize_year'] = False

yet_another_economist
```

Other ways of adding items to / specifying items of dictionaries: 

```python
the_fourth_economist = {
    'name': 'Jan Tinbergen',
    'year_of_birth': 1903,
    'year_of_death' : 1994,
    'country_of_birth': 'Netherlands',
    'nobel_prize_year': 1969
}
```

```python
# using the function dict() 
# with a list of tuples as the input/argument 
the_fifth_economist = dict([
    ('name', 'Elinor Ostrom'),
    ('year_of_birth', 1933),
    ('year_of_death', 2012),
    ('country_of_birth', 'USA'),
    ('nobel_prize_year', 2009)
])
the_fifth_economist
```
