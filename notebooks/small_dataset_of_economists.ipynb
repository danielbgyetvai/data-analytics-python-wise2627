## Functions and using `input()`
- enter data on economists using `input()` embedded in a function

```python
# string, more or less text / sequence of characters 
a_string = 'Esther Duflo'
# putting together strings
a_string = a_string + ' was born in 1972.'

a_string
```

```python
# creating the function add_economist()
def add_economist():
    # collect the value for each attribute
    # keeping the five attributes
    attribute_names = ('name', 
                       'year_of_birth', 
                       'year_of_death', 
                       'country_of_birth',
                       'nobel_prize_year')
    # empty dictionary
    # new_economist exists only within the function
    new_economist = {}
    # looping over attributes
    # for loop
    for attribute_name in attribute_names:
        # adding an element to the dictionary attribute_dict
        # with key/attribute name /variable name 
        # corresponding to the content of attribute_name 
        new_economist[attribute_name] = input(attribute_name + ': ')
    # at this point (after the for loop), we have the dictionary 
    # filled with the attribute names and values    
    # now, let us add it to the list of dictionaries we already have
    # some_economists exists outside (it was created outside)
    some_economists.append(new_economist)

```

```python
add_economist()
```

```python
class Economist:
    # initialiser function
    # to create objects of this class
    # self (convention) --> refers to the object itself
    def __init__(self):
        # five attributes
        # each at baseline / by default set to None (unknown)
        self.name = None 
        self.year_of_birth = None
        self.year_of_death = None
        self.countr_of_birth = None
        self.nobel_prize_year = None
    # add_attributes()
    # setting the attributes interactively using input()
    # function belonging to objects of the Economist class
    # jargon: called a "method" of the Economist class
    def add_attributes(self):
        # retrieve the attribute names
        # these are the keys in the __dict__ of the object itself
        attribute_names = self.__dict__.keys()
        # for each attribute name, let the user enter the corresponding value 
        # interactively using input()
        for attribute_name in attribute_names:
            attribute_value = input(attribute_name + ': ')
            # here, let us use the setattr() function
            setattr(self, attribute_name, attribute_value)


```

```python
class Economist:
    # initialiser function
    # to create objects of this class
    # self (convention) --> refers to the object itself
    def __init__(self, 
                 name = None, 
                 year_of_birth = None, 
                 year_of_death = None, 
                 country_of_birth = None,
                 nobel_prize_year = None):
        # five attributes
        # each at baseline / by default set to None (unknown)
        self.name = name
        self.year_of_birth = year_of_birth
        self.year_of_death = year_of_death
        self.country_of_birth = country_of_birth
        self.nobel_prize_year = nobel_prize_year
```

*str* = string (sequence of characters / text / qualitative)

need to address the problem causing *object* data type
in all of the cases here, the columns with object data type should be numerical / quantitative (with eventually values)

Instead of 'None' and 'False', let us have explicitly missing values `pd.NA`


```python
# selecting rows with NA
rows_with_NA = (
    df['year_of_death'].isin([None, False])
)
# changing the values to explicit NA values
df.loc[rows_with_NA, 'year_of_death'] = pd.NA

# selecting rows with NA
rows_with_NA = (
    df['nobel_prize_year'].isin([None, False])
)
# changing the values to explicit NA values
df.loc[rows_with_NA, 'nobel_prize_year'] = pd.NA



```

```python
# convert the data types 
# to pandas dtypes
df.convert_dtypes()
df.dtypes
```

in every column containing years
the values are either integer (whole number) or missing
 

```python
# setting the datatype of a column to integer 
df['year_of_birth'] = df['year_of_birth'].astype(int)
df.dtypes
```

```python
# converting data type to integer
df['year_of_death'] = (
    df['year_of_death'].astype(str).astype('Int64')
)
df.convert_dtypes()
df.dtypes
```

```python
# first convert to a string
# convert it to an integer that allows for missing values
df['nobel_prize_year'] = (
    df['nobel_prize_year']
        .astype(str)
        .astype('Int64')
)
df.convert_dtypes()
df.dtypes
```
